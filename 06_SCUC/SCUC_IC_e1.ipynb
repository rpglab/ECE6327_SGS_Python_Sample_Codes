Two-unit single-period UC.

```python
from pyomo.environ import (
    ConcreteModel, Var, Objective, Constraint, SolverFactory,
    Binary, minimize, value
)

# ---- Generator data ----
c1, SU1, Pgmin1, Pgmax1 = 10, 800, 40, 80
c2, SU2, Pgmin2, Pgmax2 = 30, 100, 20, 90
TLoad = 70

model = ConcreteModel()
model.u1 = Var(domain=Binary)
model.u2 = Var(domain=Binary)
model.Pg1 = Var()
model.Pg2 = Var()

model.obj = Objective(
    expr=(model.u1*SU1 + c1*model.Pg1) + (model.u2*SU2 + c2*model.Pg2),
    sense=minimize
)

model.PowerBalance   = Constraint(expr=model.Pg1 + model.Pg2 == TLoad)
model.genLimit_1_Min = Constraint(expr=Pgmin1*model.u1 <= model.Pg1)
model.genLimit_1_Max = Constraint(expr=model.Pg1 <= Pgmax1*model.u1)
model.genLimit_2_Min = Constraint(expr=Pgmin2*model.u2 <= model.Pg2)
model.genLimit_2_Max = Constraint(expr=model.Pg2 <= Pgmax2*model.u2)
```

```python
# ---- Solve with Gurobi ----
solver = SolverFactory('gurobi')
solver.options['MIPGap'] = 0.0
solver.options['TimeLimit'] = 90
results = solver.solve(model, tee=True)
print(results.solver.status, results.solver.termination_condition)
print(f"u1 = {value(model.u1)}, u2 = {value(model.u2)}")
print(f"Pg1 = {value(model.Pg1):.4f}, Pg2 = {value(model.Pg2):.4f}")
```

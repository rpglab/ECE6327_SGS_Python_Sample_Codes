```python
from pyomo.environ import (
    ConcreteModel, Var, Objective, Constraint, SolverFactory, maximize, value
)

model = ConcreteModel()
model.x1 = Var()
model.x2 = Var()

model.obj = Objective(expr=0, sense=maximize)
model.constName   = Constraint(expr=3*model.x1 + 2*model.x2 ==  9)
model.constName_2 = Constraint(expr=  model.x1 + 6*model.x2 == -7)
```

```python
# ---- Solve with Gurobi ----
solver = SolverFactory('gurobi')
solver.options['MIPGap'] = 0.0
solver.options['TimeLimit'] = 90
results = solver.solve(model, tee=True)
print(results.solver.status, results.solver.termination_condition)

print(f"x1 = {value(model.x1)}")
print(f"x2 = {value(model.x2)}")
```

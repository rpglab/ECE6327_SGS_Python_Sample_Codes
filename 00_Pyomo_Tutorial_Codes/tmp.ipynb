A small nonlinear feasibility system. Gurobi handles non-convex quadratic equalities when `NonConvex=2` is set.

```python
from pyomo.environ import (
    ConcreteModel, Var, Objective, Constraint, SolverFactory, maximize, value
)

model = ConcreteModel()
model.x = Var()
model.y = Var()

model.obj = Objective(expr=0, sense=maximize)

# Nonlinear (non-convex) constraints
model.constName   = Constraint(expr=model.x*model.x + 2*model.y == 3)
model.constName_2 = Constraint(expr=model.y*model.y - model.x*model.y == 1)
```

```python
solver = SolverFactory('gurobi')
solver.options['NonConvex'] = 2     # allow non-convex quadratic equalities
results = solver.solve(model, tee=True)

print(f"x = {value(model.x)}")
print(f"y = {value(model.y)}")
```

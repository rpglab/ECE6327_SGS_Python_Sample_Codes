A trivial LP: minimize a free variable subject to a two-sided bound.

```python
from pyomo.environ import (
    ConcreteModel, Var, Objective, Constraint, SolverFactory, minimize, value
)

model = ConcreteModel()

# Declare variable (free)
model.x = Var()

# Objective
model.obj = Objective(expr=model.x, sense=minimize)

# Constraint:  -1.5 <= x <= 3
model.constName = Constraint(expr=(-1.5, model.x, 3))
```

```python
# ---- Solve with Gurobi ----
solver = SolverFactory('gurobi')
solver.options['MIPGap'] = 0.0
solver.options['TimeLimit'] = 90
results = solver.solve(model, tee=True)
print(results.solver.status, results.solver.termination_condition)

print(f"x = {value(model.x)}")
```

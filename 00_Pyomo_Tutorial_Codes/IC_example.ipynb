```python
from pyomo.environ import (
    ConcreteModel, Var, Objective, Constraint, SolverFactory,
    NonNegativeReals, maximize, value
)

model = ConcreteModel()

# x1 is "free" in AMPL but a constraint x1>=0 is added separately (kept faithfully here)
model.x1 = Var()
model.x2 = Var(domain=NonNegativeReals)
model.x3 = Var(domain=NonNegativeReals)

model.obj = Objective(
    expr=2*model.x1 + 5*model.x2 + model.x3, sense=maximize
)

model.constName = Constraint(expr=  model.x1 + 3*model.x2 + 2*model.x3 <= 10)
model.uuhiih    = Constraint(expr=2*model.x1 +   model.x2 + 5*model.x3 <= 8)
model.uhoah     = Constraint(expr=model.x1 >= 0)
```

```python
solver = SolverFactory('gurobi')
results = solver.solve(model, tee=True)

print(f"x1 = {value(model.x1)}")
print(f"x2 = {value(model.x2)}")
print(f"x3 = {value(model.x3)}")
```

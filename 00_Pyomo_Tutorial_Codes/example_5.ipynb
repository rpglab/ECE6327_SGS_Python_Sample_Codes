Two-variable LP, maximize.

```python
from pyomo.environ import (
    ConcreteModel, Var, Objective, Constraint, SolverFactory,
    NonNegativeReals, maximize, value
)

model = ConcreteModel()
model.x1 = Var(domain=NonNegativeReals)
model.x2 = Var(domain=NonNegativeReals)

model.obj = Objective(expr=model.x1 + model.x2, sense=maximize)
model.constName   = Constraint(expr=3*model.x1 + 4*model.x2 <= 24)
model.constName_2 = Constraint(expr=7*model.x1 + 4*model.x2 <= 28)
```

```python
# ---- Solve with Gurobi ----
solver = SolverFactory('gurobi')
solver.options['MIPGap'] = 0.0
solver.options['TimeLimit'] = 90
results = solver.solve(model, tee=True)
print(results.solver.status, results.solver.termination_condition)

print(f"x1 = {value(model.x1)}")
print(f"x2 = {value(model.x2)}")
```
